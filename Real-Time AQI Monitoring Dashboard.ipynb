```python
import pandas as pd
import requests
import matplotlib.pyplot as plt
url = "https://api.data.gov.in/resource/3b01bcb8-0b14-4abf-b6f2-c1bfd384ba69?api-key=579b464db66ec23bdd0000010ce9da9e93e149f343f43f7a06ec3378&format=json&filters%5Bstate%5D=Bihar"
```

```python
r = requests.get(url)
```

```python
r.status_code
```

```python
data = r.json()
data["records"]
```

```python
city =[]
pollutant = []
pollutionlevel = []
updated_timestamp= []
for index,info in enumerate(data["records"]):
    city.append(info["city"])
    pollutant.append(info["pollutant_id"])
    pollutionlevel.append(info["avg_value"])
    updated_timestamp.append(info["last_update"])

df = pd.DataFrame({"Updated Time": updated_timestamp, "City" : city, "Pollutant":pollutant,"Pollution Level(Avg)":pollutionlevel})
df
```

```python
df.to_csv("./Documents/AQIrecords.csv",sep=",", mode="a",header=True, index=False)
```

```python
df
```

```python
df["Pollution Level(Avg)"] = df["Pollution Level(Avg)"].str.strip()
df["Pollution Level(Avg)"] = pd.to_numeric(df["Pollution Level(Avg)"],errors="coerce")
```

```python
df.plot(kind="bar", x = "City", y="Pollution Level(Avg)", title = "AQI Data of Differnet Cities of Bihar")
plt.ylabel(df["Updated Time"][0])
plt.tight_layout()  # Optional, adjusts layout
plt.savefig("./Documents/plot1.png")  # Save as PNG
plt.show()
```
