<h2>Basic Statistics & Summary</h2>

```python
consumer.describe()                      # Numeric summary
consumer['PAY_MODE'].value_counts()     # Frequency of categories
# consumer['TARIFF'].unique()             # Unique tariff codes
```

<h2>Distribution of Amount Paid</h2>

```python
sns.histplot(consumer['AMOUNT'], bins=10, kde=True)
plt.title("Distribution of Paid Amounts")
plt.show()
```

<h2>Comparison of Paid vs Billed Amount</h2>

```python
sns.scatterplot(x='BILLED_AMOUNT', y='AMOUNT', data=consumer)
plt.title("Paid Amount vs Billed Amount")
plt.show()
```

<h2>Payment Mode Analysis</h2>

```python
sns.boxplot(x='PAY_MODE', y='AMOUNT', data=consumer)
plt.title("Payment Amount by Payment Mode")
plt.show()
```

```python
consumer.set_index('COLL_DT')['AMOUNT'].resample('D').sum().plot(figsize=(12,6))
plt.title("Daily Revenue Collected")
plt.ylabel("Amount")
plt.show()
```

<h2>Monthly Aggregation</h2>

```python
consumer['Month'] = consumer['COLL_DT'].dt.to_period('M')
monthly_revenue = consumer.groupby('Month')['AMOUNT'].sum()
monthly_revenue.plot(kind='bar', figsize=(14,6))
plt.title("Monthly Revenue Collection")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.show()
```

<h2>Revenue by Tariff and Payment Mode</h2>

```python
pivot = pd.pivot_table(consumer, values='AMOUNT', index='TARIFF', columns='PAY_MODE', aggfunc='sum')
pivot.plot(kind='bar', stacked=True, figsize=(12,6))
plt.title("Revenue by Tariff and Payment Mode")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.show()
```
